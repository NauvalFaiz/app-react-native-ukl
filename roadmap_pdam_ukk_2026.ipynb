```mermaid
graph TD
    %% --- SEKTOR AWAL & SPLASH SCREEN ---
    Start([Mulai Aplikasi]) --> Splash[Halaman Splash Screen]
    Splash --> ChkAppKey{Apakah APP-KEY / \nOwner Token sudah ada\n di file config/.env?}
    
    ChkAppKey -- Belum / Pertama Kali --> Postman[Buka Postman: POST /app-owners]
    Postman --> SaveEnv[Masukkan token ke config/.env perangkat]
    SaveEnv --> Splash
    
    ChkAppKey -- Sudah Ada --> ChkToken{Apakah JWT Token\nTersimpan di Storage?}

    %% --- SEKTOR AUTENTIKASI ---
    ChkToken -- Tidak Ada --> LoginView[Halaman Login]
    ChkToken -- Ada / Sesi Aktif --> HitMe[Hit API: GET /me]
    
    LoginView --> BtnRegAdmin{Klik Tombol\nRegister Admin?}
    BtnRegAdmin -- Ya --> RegAdminView[Halaman Register Admin]
    RegAdminView --> HitReg[Hit API: POST /admins]
    HitReg -- Sukses --> LoginView
    
    BtnRegAdmin -- Tidak --> InputAuth[Input Username & Password]
    InputAuth --> HitAuth[Hit API: POST /auth]
    
    HitAuth -- Gagal / 401 --> AlertAuth[Tampilkan 'Username/Password Salah']
    AlertAuth --> LoginView
    
    HitAuth -- Berhasil / 200 --> SaveToken[Simpan JWT Token & Role\nke Local Storage / SharedPref]
    SaveToken --> HitMe

    %% --- SEKTOR CEK ME & PENENTUAN ROLE ---
    HitMe -- Token Expired / Gagal --> ClearToken[Hapus Token dari Storage]
    ClearToken --> LoginView
    
    HitMe -- Berhasil / 200 --> RoleCheck{Cek Parameter 'role'}

    %% --- SEKTOR UTAMA: ROLE ADMIN ---
    RoleCheck -- "ADMIN" --> AdminDash[Dashboard Utama Admin]
    
    %% Admin - Modul Layanan
    AdminDash --> MenuS[Menu Kelola Layanan]
    MenuS --> GetS[GET /services] --> ListS[Tampilkan List Layanan]
    ListS --> ActionS{Aksi Layanan?}
    ActionS -- Tambah --> PostS[POST /services] --> GetS
    ActionS -- Edit --> PutS[PUT /services/:id] --> GetS
    ActionS -- Hapus --> DelS[DELETE /services/:id] --> GetS
    
    %% Admin - Modul Customer
    AdminDash --> MenuC[Menu Kelola Pelanggan]
    MenuC --> GetC[GET /customers] --> ListC[Tampilkan List Pelanggan]
    ListC --> ActionC{Aksi Pelanggan?}
    ActionC -- Tambah --> FetchDropdownS[Jalankan GET /services untuk isi Dropdown]
    FetchDropdownS --> PostC[POST /customers] --> GetC
    ActionC -- Edit --> PutC[PUT /customers/:id] --> GetC
    ActionC -- Hapus --> DelC[DELETE /customers/:id] --> GetC
    
    %% Admin - Modul Tagihan (Bills)
    AdminDash --> MenuB[Menu Input Tagihan]
    MenuB --> FetchDropdownC[Jalankan GET /customers untuk List]
    FetchDropdownC --> FormB[Input Meteran Bulanan]
    FormB --> PostB[POST /bills] --> AlertB[Tagihan Berhasil Dibuat]
    
    %% Admin - Modul Verifikasi (Payments)
    AdminDash --> MenuV[Menu Verifikasi Bayar]
    MenuV --> GetV[GET /payments] --> ListV[Tampilkan Foto Bukti dari Customer]
    ListV --> PutV[Aksi Approve: PUT /payments/:id] --> GetV
    
    %% Admin - Profil & Logout
    AdminDash --> ProfileA[Halaman Profil Admin]
    ProfileA --> LogoutA[Klik Logout]

    %% --- SEKTOR UTAMA: ROLE CUSTOMER ---
    RoleCheck -- "CUSTOMER" --> CustDash[Dashboard Utama Customer]
    
    %% Customer - Grafik & Tagihan
    CustDash --> FetchCustData[Hit API: GET /bills/me]
    FetchCustData --> RenderUI[1. Hitung Jumlah Tagihan paid=false\n2. Masukkan data ke Grafik Batang]
    
    %% Customer - Bayar Tagihan
    CustDash --> MenuPay[Halaman Tagihan Saya]
    MenuPay --> SelectBill[Pilih Tagihan yang Belum Dibayar]
    SelectBill --> PickImage[Buka Galeri HP / Kamera]
    PickImage --> PostPay[Hit API multipart/form-data: POST /payments]
    PostPay --> AlertPay[Bukti Berhasil Dikirim, Menunggu Verifikasi]
    
    %% Customer - Profil & Logout
    CustDash --> ProfileC[Halaman Profil Customer]
    ProfileC --> LogoutC[Klik Logout]

    %% --- SEKTOR LOGOUT KELUAR AKUN ---
    LogoutA --> DoLogout[Hapus JWT Token dari Local Storage]
    LogoutC --> DoLogout
    DoLogout --> RedirectLogin[Arahkan kembali ke Halaman Login]
    RedirectLogin --> LoginView

    %% --- STYLING DIAGRAM ---
    style Start fill:#2ecc71,stroke:#27ae60,stroke-width:2px,color:#fff
    style LoginView fill:#3498db,stroke:#2980b9,stroke-width:2px,color:#fff
    style AdminDash fill:#9b59b6,stroke:#8e44ad,stroke-width:2px,color:#fff
    style CustDash fill:#e67e22,stroke:#d35400,stroke-width:2px,color:#fff
    style RoleCheck fill:#f1c40f,stroke:#f39c12,stroke-width:2px,color:#000
    style Postman fill:#e74c3c,stroke:#c0392b,stroke-width:2px,color:#fff
```
